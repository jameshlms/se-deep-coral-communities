```python
import pandas as pd
```

```python
def _print_dims(df: pd.DataFrame) -> None:
    r, c = df.shape
    print(f"DataFrame shape: {r} rows, {c} columns")
```

```python
df = pd.read_parquet("../data/interim/se_coral_records.parquet")
df.head(10)
```

```python
observation_dates = pd.to_datetime(
    df["observation_date"],
    format="%Y-%m-%d",
    errors="coerce",
)

df["observation_date"] = observation_dates
df = df.dropna(subset=["observation_date"])

_print_dims(df)
```

```python
df = df.replace("", None, inplace=True)

df = df[df["family"].notna() | df["genus"].notna() | df["order"].notna()]

_print_dims(df)
```

```python
df.to_parquet("../data/processed/se_coral_records_cleaned.parquet", index=False)
```
